```python

import numpy as np
from PIL import Image
from io import BytesIO
from urllib import request

def download_image(url):
    with request.urlopen(url) as resp:
        buffer = resp.read()
    stream = BytesIO(buffer)
    img = Image.open(stream)
    return img

def prepare_image(img, target_size=(200,200)):
    if img.mode != 'RGB':
        img = img.convert('RGB')
    img = img.resize(target_size, Image.NEAREST)
    return img

def preprocess(img):
    arr = np.array(img).astype(np.float32) / 255.0
    mean = np.array([0.485, 0.456, 0.406])
    std  = np.array([0.229, 0.224, 0.225])
    arr = (arr - mean) / std
    return arr

url = "https://habrastorage.org/webt/yf/_d/ok/yf_dokzqy3vcritme8ggnzqlvwa.jpeg"
img = download_image(url)
img = prepare_image(img, target_size=(200,200))
arr = preprocess(img)


first_pixel_r = arr[0,0,0]
print("First pixel R channel after preprocessing:", first_pixel_r)


```

```python
# Step 0: Install dependencies
!pip install onnxruntime pillow requests

# Step 1: Download the prebuilt ONNX model (replace with course URL if needed)
import requests

model_url = "https://github.com/alexeygrigorev/large-datasets/releases/download/hairstyle/hair_classifier_v1.onnx"
model_path = "hair_classifier_empty.onnx"  # locally save as empty for consistency

resp = requests.get(model_url)
with open(model_path, "wb") as f:
    f.write(resp.content)
print("Model downloaded:", model_path)

# Step 2: Preprocessing (matches HW8)
from PIL import Image
import numpy as np

TARGET_SIZE = (200, 200)
MEAN = np.array([0.485, 0.456, 0.406], dtype=np.float32)
STD = np.array([0.229, 0.224, 0.225], dtype=np.float32)

def preprocess_image(img):
    if img.mode != "RGB":
        img = img.convert("RGB")
    img = img.resize(TARGET_SIZE, Image.NEAREST)
    arr = np.array(img).astype(np.float32) / 255.0
    arr = (arr - MEAN) / STD
    arr = arr.transpose(2, 0, 1)
    arr = np.expand_dims(arr, 0)
    arr = arr.astype(np.float32)  # ensure float32
    return arr


# Step 3: Load test image
img_url = "https://habrastorage.org/webt/yf/_d/ok/yf_dokzqy3vcritme8ggnzqlvwa.jpeg"
resp = requests.get(img_url)
img = Image.open(BytesIO(resp.content))

# Optional: display
img.show()

# Step 4: Preprocess image
input_arr = preprocess_image(img)

# Step 5: Run model
import onnxruntime as rt

sess = rt.InferenceSession(model_path, providers=['CPUExecutionProvider'])
input_name = sess.get_inputs()[0].name
output_name = sess.get_outputs()[0].name

prob = sess.run([output_name], {input_name: input_arr})
print("Predicted probability:", float(prob[0][0][0]))
```
